## Phenotype Data
- Generates all necessary phenotype .csv files

```python
import pandas as pd
import numpy as np
```

```python
fphenotype = "../data/phenotype/phenotype_matrix.xlsx"
```

```python
#
# only a single sheet 
#
excel_p = pd.ExcelFile(fphenotype)
excel_p.sheet_names
```

```python
df_p = pd.read_excel(excel_p)
# df_p.head()
```

#### Data Processing

```python
#
# make all float
#
df_p.iloc[:, 1:] = df_p.iloc[:, 1:].astype(np.float64)
# df_p.iloc[:20,:]
```

```python
#
# Since one is interested in the significant markers, one can clearly 
# replace NaN's with the colums mean. Since the number of samples is not 
# not to great, dropping might has a noticable statistical effect here: 
# -> df_p.dropna(inplace=True) | alternative
#
for i in range(1, len(df_p.columns)):
    df_p.iloc[:, i].replace(
        to_replace=np.float64("nan"), 
        value=df_p.iloc[:, i].mean(skipna=True), 
        inplace=True
    )
# df_p.iloc[:20,:]
```

```python
#
# export
#
df_p.to_csv("../data/phenotype/phenotype_matrix.csv", index=False)
```
